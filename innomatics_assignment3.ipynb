```python
#Problem Statement 1: Employee Performance Bonus Eligibility

employees = {
"Ravi": 92,
"Anita": 88,
"Kiran": 92,
"Suresh": 85
} #this input is given already
maxscore = max(employees.values()) #now find the highest score in the group
topperform = [name for name, score in employees.items() if score == maxscore] #Find everyone who reached that score
print(f"Top Performers Eligible for Bonus: {', '.join(topperform)} (Score: {maxscore})")
```

```python
#Problem Statement 2: Search Query Keyword Analysis

query = "Buy mobile phone buy phone online"
# lowercase and split
words = query.lower().split()

# count occurrences
counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

# keep only duplicates
duplicates = {word: count for word, count in counts.items() if count > 1}
print(duplicates)
```

```python
#Problem Statement 3: Sensor Data Validation

sensor_readings = [3, 4, 7, 8, 10, 12, 5]
# Using enumerate to keep track of the 'Hour' (the list index)
valid = [(hour, val) for hour, val in enumerate(sensor_readings) if val%2==0]
print(f"Valid Sensor Readings (Hour, Value):\n {valid}")
```

```python
#Problem Statement 4: Email Domain Usage Analysis

emails = [
    "ravi@gmail.com",
    "anita@yahoo.com",
    "kiran@gmail.com",
    "suresh@gmail.com",
    "meena@yahoo.com"
]
#collect only the domains
domains = []
for mail in emails:
    at_pos = mail.find('@')        # find position of '@'
    domain_part = mail[at_pos + 1:]   # take text after '@'
    domains.append(domain_part)
#count the frequency manually
domain_count = {}
for d in domains:
    if d not in domain_count:
        domain_count[d] = 1
    else:
        domain_count[d] += 1
#calculate the percentage now
total_users = len(emails)
for d in domain_count:
    percent = (domain_count[d] / total_users) * 100
    print(f"{d}: {percent:.0f}%")
```

```python
#Problem Statement 5: Sales Spike Detection

# Sales Spike Detection Program
sales = [1200, 1500, 900, 2200, 1400, 3000]
average_sale = sum(sales) / len(sales)
for i in range(len(sales)):
    increase_percent = ((sales[i] - average_sale) / average_sale) * 100
    if increase_percent >= 29:
        print(f"Day {i+1}: {sales[i]}")


```

```python
#Problem Statement 6: Duplicate User ID Detection

user_ids = ["user1", "user2", "user1", "user3", "user1", "user3"]
seen = set()          # stores first occurrences
duplicates = {}       # stores repeated ids
for uid in user_ids:
    if uid in seen:
        duplicates[uid] = duplicates.get(uid, 1) + 1
    else:
        seen.add(uid)
# printing result
for uid, count in duplicates.items():
    print(f"{uid} → {count} times")

```
